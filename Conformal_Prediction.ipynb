```python

# conformal prediction
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# Load data
X, y = load_diabetes(return_X_y=True)

# Three-way split: proper-train / calibration / test
X_traincal, X_test, y_traincal, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0
)
X_train, X_cal, y_train, y_cal = train_test_split(
    X_traincal, y_traincal, test_size=0.33, random_state=0
)

X_train.shape, X_cal.shape, X_test.shape
```

```python
model = LinearRegression()
model.fit(X_train, y_train)

# calibration residuals
y_cal_pred = model.predict(X_cal)
scores = np.abs(y_cal - y_cal_pred)
n_cal = len(scores)

print(f"n_cal = {n_cal}, mean calibration abs residual = {scores.mean():.3f}")
```
