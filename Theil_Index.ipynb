```python
# 21-07-26
import numpy as np

def calculate_theil_fairness(x, epsilon=1e-8):
    """
    Calculates the Theil's T Index for individual fairness.
    x: Array of individual benefits (predictions or probabilities).
    """
    x = np.clip(np.array(x, dtype=float), epsilon, None)
    mu = np.mean(x)

    if mu == 0:
        return 0.0

    return float((1 / len(x)) * np.sum((x / mu) * np.log(x / mu)))

# Example execution, higher value means more inequality
# predictions = [1,1,1,0,0,1,0,1,1,1]
predictions = [1,0,1,0,0,1,0,0,1,0] #benefit is correct and favourable
print(f"Individual Unfairness Score: {calculate_theil_fairness(predictions):.4f}")
```
